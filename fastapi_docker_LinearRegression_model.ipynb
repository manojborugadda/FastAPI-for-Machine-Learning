```python
%pip install scikit-learn pandas numpy
import pandas as pd
import numpy as np
```

```python
num_datapoints = 100
```

```python
X = np.random.rand(num_datapoints, 2)
print(X)
```

```python
weights = np.array([2,3])
```

```python
bias = 5
```

```python
y = X.dot(weights) + bias + np.random.randn(num_datapoints)
```

```python
y
```

```python
# we got X and Y now and we will do the
# creation of LINEAR REGRESSION MODEL
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.coef_
print(f'constant is {model.coef_[0]} and coefficient is {model.coef_[1]}')
```

```python
# y = m * x  + c
y = 3.442148881575524 * X + 2.3513290265094757
print(y)
```

```python
# saving the model and loading the models using JOBLIB
import joblib
joblib.dump(model,"linear_regression_model.pkl")
```

```python
modelTest = joblib.load("linear_regression_model.pkl")
```
